# What is the average yield of each variety of crop?



```python
import seaborn as sns
import matplotlib.pyplot as plt

avg_yield = df.groupby('Crop_Type')['Crop_Yield_MT_per_HA'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(x='Crop_Type', y='Crop_Yield_MT_per_HA', data=avg_yield)
plt.title('Average Yield per Crop Type')
plt.xlabel('Crop Type')
plt.ylabel('Average Yield (MT/HA)')
plt.xticks(rotation=45)
plt.show()
```

# How can crop yield get affected by extreme weather?

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
corr = df[['Extreme_Weather_Events', 'Crop_Yield_MT_per_HA']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Between Extreme Weather Events and Crop Yield')
plt.show()
```

```python
corr
```

# How does the application of fertilizer affect crop yield?

```python
sns.jointplot(x='Fertilizer_Use_KG_per_HA', y='Crop_Yield_MT_per_HA', data=corr, kind='hex', color='orange')
plt.suptitle('Fertilizer Use vs. Crop Yield', y=1.02)
plt.xlabel('Fertilizer Use (KG/HA)')
plt.ylabel('Crop Yield (MT/HA)')
plt.show()
```

# 4.What economic impact does climate change have on agriculture?

```python
sns.pairplot(df[['Average_Temperature_C', 'Total_Precipitation_mm', 'Extreme_Weather_Events', 'Economic_Impact_Million_USD']])
plt.title('Climate Variables vs Economic Impact')
plt.show()
```

# 5. Which region has experienced the highest average crop yield over the years?

```python
avg_yield_region = df.groupby('Region')['Crop_Yield_MT_per_HA'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(x='Crop_Yield_MT_per_HA', y='Region', data=avg_yield_region, palette='viridis')
plt.title('Average Yield by Region')
plt.xlabel('Average Yield (MT/HA)')
plt.ylabel('Region')
plt.show()
```

```python
avg_yield_region_year = df.groupby(['Region', 'Year'])['Crop_Yield_MT_per_HA'].mean().reset_index()
plt.figure(figsize=(12, 8))
sns.lineplot(x='Year', y='Crop_Yield_MT_per_HA', hue='Region', data=avg_yield_region_year, marker='o')
plt.title('Average Crop Yield by Region Over the Years')
plt.xlabel('Year')
plt.ylabel('Average Crop Yield (MT/HA)')
plt.legend(title='Region', loc='upper left')
plt.xticks(rotation=90)
plt.show()
```

# 6. How does the average temperature relate to the amount of crops produced?

```python
sns.pairplot(df[['Average_Temperature_C', 'Total_Precipitation_mm', 'Crop_Yield_MT_per_HA']])
plt.suptitle('Temperature, Precipitation, and Crop Yield Relationship', y=1.02)
plt.show()
```

# 7.How does CO2 emissions affect crop yield in different regions?

```python
plt.figure(figsize=(10, 8))
corr = df[['CO2_Emissions_MT','Crop_Yield_MT_per_HA','Average_Temperature_C','Total_Precipitation_mm','Extreme_Weather_Events']].corr()
sns.heatmap(corr, annot=True, cmap='RdYlGn', vmin=-1, vmax=1)
plt.title('Correlation Between CO2 Emissions, Crop Yield, and Environmental Factors')
plt.show()
```

# 8.Which types of crops are more resilient to severe weather conditions?

```python
threshold = 5  
extreme_weather = df[df['Extreme_Weather_Events'] > threshold]

plt.figure(figsize=(10,6))
sns.barplot(y='Crop_Type', x='Crop_Yield_MT_per_HA', data=extreme_weather)
plt.title('Crop Yield Resilience During Extreme Weather')
plt.xlabel('Crop Type')
plt.ylabel('Crop Yield (MT/HA)')
plt.xticks(rotation=45)
plt.show()
```

# . Which region has the best soil health, and how does it affect pesticide use?

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='Soil_Health_Index', y='Pesticide_Use_KG_per_HA', hue='Region', data=df, palette='Set2')
plt.title('Soil Health vs Pesticide Use by Region')
plt.xlabel('Soil Health Index')
plt.ylabel('Pesticide Use (KG/HA)')
plt.show()
```

# 10.How do adaptation strategies influence crop yield in regions with frequent extreme weather?

```python
df_extreme = df[df['Extreme_Weather_Events'] > threshold]
avg_yield_strategy = df_extreme.groupby('Adaptation_Strategies')['Crop_Yield_MT_per_HA'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(x='Adaptation_Strategies', y='Crop_Yield_MT_per_HA', data=avg_yield_strategy)
plt.title('Adaptation Strategies vs Crop Yield in Extreme Weather Regions')
plt.xlabel('Adaptation Strategies')
plt.ylabel('Average Crop Yield (MT/HA)')
plt.xticks(rotation=90)
plt.show()
```
