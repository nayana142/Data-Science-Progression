```python
# import some basic libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('Churn_Modelling.csv')
```

```python
## Divide the dataset into independent and dependent features
here first 3 columns are unwanted(row_number,customer_id,first_name), so our X becomes
X = dataset.iloc[:,3:13]
Y = dataset.iloc[:,13]----it is a binary class classification problem(0 or 1)
```

```python
## convert categorical features into integers(using onehot encoding)
geography=pd.get_dummies(X['Geography'],drop_first=True)
gender=pd.get_dummies(X['Gender'],drop_first=True)


## concatenate these variables with dataframe
X.drop(['Geography','Gender'],axis=1)

X=pd.concat([X,geography,gender],axis=1)
```

```python
# splitting the dataset into training and testing

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
## Feature scaling

from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

# Part 2 : CREATE THE ANN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import ReLU,LeakyReLU,PReLU,ELU
from tensorflow.keras.layers import Dropout
```

```python
## initialize the ANN

classifier=Sequential()
```

```python
## adding the input layer

classifier.add(Dense(units=11,activation='relu'))


# if we want to add dropout layer then
classifier.add(Dense(units=11,activation='relu'))
classifier.add(Dropout(0.2))  #20 % of input features will dead
```

```python
## adding the first hidden layer

classifier.add(Dense(units=7,activation='relu'))
```

```python
## adding the second hiddden layer

classifier.add(Dense(units=6,activation='relu'))
```

```python
## adding the output layer

classifier.add(Dense(units=1,activation='sigmoid'))
```

```python
classifier.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
## by default adam using learning rate = 0.1
# if we want to use another learning rate 
import tensorflow
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
classifier.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
## training our model

model_history=classifier.fit(x_train,y_train,validation_split=0.33,batch_size=10,epochs=1000)
```

```python
# after completing some epochs our accuracy remains constant. then we use early stopping.
# the kernal automatically stop if the accuracy is not increases .
# stop training when a monitor metric has stopped improving

import tensorflow as tf
early_stopping=tf.keras.callbacks.EarlyStopping(
                monitor='val_loss',
                min_delta=0.0001,
                patience=20,
                verbose=1,
                mode='auto',
                baseline=None,
                restore_best_weights=False,
                )
```

```python
model_history=classifier.fit(x_train,y_train,validation_split=0.33,batch_size=10,epochs=1000,callbacks=early_stopping)
```

# Part 3 : Predictions and evaluating the model

```python
# predicting the test set results

y_pred=classifier.predict(x_test)
y_pred=(y_pred >= 0.5)

# if y_pred >= 0.5 then output=1 else output=0
```

```python
## make the confusion matrix

from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
cm
```
